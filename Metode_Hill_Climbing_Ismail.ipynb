```python
import random

# Fungsi objektif yang akan dimaksimalkan
def objective_function(x):
    # Contoh fungsi objektif: f(x) = -x^2 + 4x
    return -x**2 + 4*x

# Algoritma Hill Climbing
def hill_climbing():
    # Inisialisasi solusi secara acak
    current_solution = random.uniform(-10, 10)
    current_value = objective_function(current_solution)

    step_size = 0.1
    max_iterations = 1000

    for _ in range(max_iterations):
        # Buat tetangga solusi
        neighbors = [current_solution + step_size, current_solution - step_size]

        # Evaluasi tetangga
        next_solution = max(neighbors, key=objective_function)
        next_value = objective_function(next_solution)

        # Cek apakah solusi lebih baik
        if next_value > current_value:
            current_solution = next_solution
            current_value = next_value
        else:
            # Jika tidak ada solusi lebih baik, berhenti
            break

    return current_solution, current_value

# Contoh penggunaan
best_solution, best_value = hill_climbing()
print(f"Best solution: {best_solution}")
print(f"Best value: {best_value}")
```

Contoh studi kasus baru mencari nilai terbesar dalam Rentang -10 hingga 10 dengan menggunakan metode Hill Climbing berikut adalah codingannya :

```python
import random

# Fungsi objektif: f(x) = -x^2 + 10
def objective_function(x):
    return -x**2 + 10  # Nilai maksimum di x = 0

# Algoritma Hill Climbing
def hill_climbing():
    # Mulai dari nilai acak antara -10 hingga 10
    current_x = random.uniform(-10, 10)
    current_value = objective_function(current_x)

    step_size = 0.5  # Langkah pergerakan
    max_iterations = 1000

    for _ in range(max_iterations):
        # Coba dua kemungkinan: langkah ke kanan (+step_size) atau ke kiri (-step_size)
        neighbors = [current_x + step_size, current_x - step_size]

        # Pilih solusi terbaik dari tetangga
        next_x = max(neighbors, key=objective_function)
        next_value = objective_function(next_x)

        # Jika nilai baru lebih baik, perbarui solusi
        if next_value > current_value:
            current_x = next_x
            current_value = next_value
        else:
            # Jika tidak ada perbaikan, berhenti
            break

    return current_x, current_value

# Contoh penggunaan
best_x, best_value = hill_climbing()
print(f"Best x: {best_x}")
print(f"Best value: {best_value}")
```
